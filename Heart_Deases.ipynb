```python
# Import Library

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split,cross_validate,cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
# Loading the dataset
df = pd.read_csv("heart_disease.csv")
df.head()
```

```python
# basic info

df.info()
```

```python
# find null values

df.isnull().sum()
```

```python
# duplicat row

df.duplicated().sum()
```

```python
x = df.drop("target",axis = 1)
y = df["target"]
```

```python
# Splitting the dataset for training and testing

x_train ,x_test ,y_train ,y_test = train_test_split(
    x,y,test_size = 0.3,random_state = 40
)
```

```python
print("Training Data shape : ", x_train.shape)
print("Testing Data shape : ", x_test.shape)
```

```python
# Model Training
LR = LogisticRegression(max_iter=1000)
LR.fit(x_train ,y_train)
```

```python
# Testing Accuracy
y_pred = LR.predict(x_test)
accuracy = accuracy_score(y_pred, y_test)
print("Testing Accuracy : ", accuracy)
```

```python
# Traning Accuracy
y_pred_train = LR.predict(x_train)
accuracy_train = accuracy_score(y_pred_train, y_train)
print("Testing Accuracy : ", accuracy_train)
```

```python
# validation accuracy

val_accuracy = cross_val_score(
    LR,x_train ,y_train ,cv=5
)

print("Validation Accuracy : ",val_accuracy.mean())
```

```python
# Random Forest Clasiiifier
RF = RandomForestClassifier(
    n_estimators = 200,max_depth = 10,min_samples_leaf=10,verbose=1,n_jobs=1
)
```

```python
RF.fit(x_train,y_train)
```

```python
# Testing Accuracy of Random Forest
y_pred_rf = RF.predict(x_test)
accuracy_rf = accuracy_score(y_pred_rf, y_test)
print("Testing Accuracy : ", accuracy_rf)
```

```python
# Traning Accuracy of Random Forest
y_pred_train_rf= RF.predict(x_train)
accuracy_train_rf = accuracy_score(y_pred_train_rf, y_train)
print("Testing Accuracy : ", accuracy_train_rf)
```

```python
# GradientBoostingClassifier

gb = GradientBoostingClassifier(
    n_estimators=200 , min_samples_leaf= 10 ,learning_rate=0.1 ,random_state=42
)

gb.fit(x_train ,y_train)
```

```python
# Testing Accuracy of Gardient Boosting
y_pred_gb =gb.predict(x_test)
accuracy_gb = accuracy_score(y_pred_gb, y_test)
print("Testing Accuracy : ", accuracy_gb)
```

```python
# Traning Accuracy of Gardient Boosting
y_pred_train_gb= gb.predict(x_train)
accuracy_train_gb = accuracy_score(y_pred_train_gb, y_train)
print("Testing Accuracy : ", accuracy_train_gb)
```

```python
valid_gb = cross_val_score(
    gb,x_train,y_train ,cv = 5
)

print("Validation Accuracy : ", valid_gb.mean())
```

```python
confusion_matrix(y_pred_gb,y_test)
```
